## 1. My lane as an ML task (type)

**Classification**. We are answering the question: 'Will this piece of content decline in traffic?'. We want to classify pages into 'declining' vs 'not declining' so that an editor knows which ones to review and refresh.

```python
task_type = 'classification'
print(f'Task Type: {task_type}')
```

## 2. Target or proxy

The target is a yes/no label indicating if the content is declining. In a real warehouse setting, this would be an **observed outcome** in a future 30-day window (e.g., traffic drops by >20% compared to the previous 30 days). For our starter dataset proxy, we will use the `is_declining_label` (which is derived from `trend_direction == \'down\'`).

```python
target_column = 'is_declining_label'
print(f'Target: {target_column}')
```

## 3. Success metric

For classification with imbalanced or asymmetric costs, **ROC-AUC** and **Precision/Recall** (specifically precision at a certain recall, or precision vs base rate) are the right metrics. If an editor can only fix 50 pages, we care deeply about precision among the top 50 predictions (Precision@K).

```python
primary_metric = 'ROC-AUC'
secondary_metric = 'Precision@K'
print(f'Metrics: {primary_metric}, {secondary_metric}')
```

## 4. The unit of analysis, as a real dataframe

The unit of analysis is **one content item for a specific client** (represented by `content_id` and `client_id`). Each row aggregates the performance of one page over a 90-day window.

```python
import pandas as pd
df = pd.read_csv('../../data/raw/content_refresh_anonymized.csv')
print(f'Dataset shape: {df.shape}')
print('One row = one content item (content_id) per client (client_id)')
display(df[['content_id', 'client_id', 'content_type', 'impressions_90d', 'clicks_90d', 'trend_direction']].head(3))
```

## 5. Why ML beats a fixed rule here

A fixed rule (e.g., 'if age > 180 days and CTR < 2% then flag for refresh') is too rigid. Decline is a complex, tangled pattern involving content age, shifting search intent, competitive levels, word count, and engagement rates. ML can weigh these many weak signals together dynamically, saving editor hours that would be wasted on false positives from simple IF-statements.

```python
reasons_ml_wins = [
    'Many weak signals (age, intent, word count, engagement) interact in non-linear ways',
    'Rules degrade over time and require manual tuning',
    'ML provides a probability score, allowing editors to rank the urgency of fixes'
]
for i, reason in enumerate(reasons_ml_wins, 1):
    print(f'{i}. {reason}')
```
